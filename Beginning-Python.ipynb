```python
import pandas as pd

banka_verileri = {
    "Ad": ["Burak", "Ali", "Elif", "Zeynep", "Can"],
    "Numara": [2031234, 5912301, 8240273, 1203945, 9876543],
    "Bakiye": [1500.75, 2500.00, 3200.50, 1800.25, 4000.00],
    "Tur": ["Vadesiz", "Vadeli", "Vadesiz", "Vadeli", "Yatırım"],
    "Aktif_Mi": [True, True, False, True, False],
}

df_banka = pd.DataFrame(banka_verileri)
#print(df_banka)
print(df_banka[df_banka["Bakiye"] > 2000])
print("\n")
print(df_banka[df_banka["Tur"] == "Vadesiz"])
print("\n")
print(df_banka[(df_banka["Bakiye"] > 2000) & (df_banka["Tur"] == "Vadesiz")])
df_banka["Zam"] = df_banka["Bakiye"] + (df_banka["Bakiye"] * 0.2)
print("\n")
print("Zam yapılmış halleri:")
sonuc = (df_banka["Ad"] + " - " + df_banka["Zam"].astype(str))
print(sonuc)
```

MIDTERM PREPARATION


```python
array = [1, -1, 0]

for x in array:
    if x < 0:
        print("x = {} is strictly negative.".format(x))
    elif x > 0:
        print("x = {} is strictly positive.".format(x))
    else:
        print("x = {}".format(x))
print("Program ends.")
```

```python
str_list = ['saw', 'small', 'He', 'foxes', [123], 'six']
print(str_list)
str_list.sort(key=len)  
print(str_list)  # Output: [[123], 'He', 'six', 'saw', 'small', 'foxes']
```

```python
students = ['bob', 'alice', 'carl']

print("Here is the list in alphabetical order:")
print(sorted(students))  # Output: ['alice', 'bob', 'carl']

print("Here is the list in reverse alphabetical order:")
print(sorted(students, reverse=True))  # Output: ['carl', 'bob', 'alice']

print(students)  # The original list remains unchanged
```
